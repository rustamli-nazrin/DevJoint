## Datanın Yüklənməsi və İlkin Baxış

```
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import kagglehub
import os
import warnings
warnings.filterwarnings("ignore")
```

```
# Kaggle-dan dataset-in ən son versiyasının endirilməsi
path = kagglehub.dataset_download("sehriyarmemmedli/binaaz-sale-project")

# Qovluqdakı CSV faylının tapılması və oxunması
csv_file = [f for f in os.listdir(path) if f.endswith('.csv')][0]
df = pd.read_csv(os.path.join(path, csv_file))
```

```
# ilk 10 sətirin göstərilməsi
df.head(10)
```

```
# Datanın ümumi sütun tipləri və boş olmayan dəyərlərin sayını yoxlamaq
df.info()
```

```
# Sətir və sütun sayı
df.shape
```

```
# Hər sütun üzrə boş dəyərlərin sayı
missing = pd.DataFrame({
    "Missing Values": df.isnull().sum(),
    "Percentage": (df.isnull().sum()/len(df))*100
})

missing.sort_values("Percentage", ascending=False)
```

##2) Boş dəyərlərin hər sütun üzrə əsaslandırılmış strategiya ilə idarə olunması

```
# bəzi sütunlarda boşluq çox yüksək olduğundan onları silirik.
drop_columns = [
    "Binanın növü",
    "featured",
    "vip",
    "hour_y"
]

df = df.drop(columns=drop_columns)
```

```
# Dublikat sətirlərin sayı
df.duplicated().sum()
```

```
# Kateqorik sütunlardakı boşluqları "Məlum deyil" ilə əvəzləyirik
categorical_cols = df.select_dtypes(include=['object']).columns
for col in categorical_cols:
    df[col] = df[col].fillna("Unknown")
```

```
# Ədədi sütunlardakı boşluqları median ilə doldururuq
# median seçirəm çünki o kənar qiymətlərə (outlier) qarşı həssasdır.
numerical_cols = df.select_dtypes(include=["int64","float64"]).columns

for col in numerical_cols:
    df[col] = df[col].fillna(df[col].median())
```

```
# yeniden yoxlayıram
df.isnull().sum()
```

##3) Outlier-lərin aşkarlanması

```
# Numeric sütunlara yenidən baxırıq.
df.describe()
```

```
# Qiymət və sahə sütunu üzrə outlier-ləri aşkarlayırıq.
# Qiymət dəyişəni sağa meyilli paylanmaya malik olduğundan İQR üsulu seçildi.
for col in ['total_price', 'Sahə']:
    if col in df.columns:
        # str tipindəki sütunları rəqəmə çevirdim məs. '120 m²' -> 120)
        if df[col].dtype == 'object':
            df[col] = df[col].str.replace(r'[^\d.]', '', regex=True).astype(float)

        Q1 = df[col].quantile(0.25)
        Q3 = df[col].quantile(0.75)
        IQR = Q3 - Q1

        lower_bound = Q1 - 1.5 * IQR
        upper_bound = Q3 + 1.5 * IQR

        outliers_count = len(df[(df[col] < lower_bound) | (df[col] > upper_bound)])
        print(f"{col} sütunundakı outlier sayı: {outliers_count}")


        df = df[(df[col] >= lower_bound) & (df[col] <= upper_bound)]

print(f"Outlier-lər təmizləndikdən sonra yeni ölçü: {df.shape}")
```

##4) Feature engineering

```
# Sütun 1: Kvadratmetrin qiyməti.əmlakın real dəyərini analiz etmək üçün.
if 'total_price' in df.columns and 'Sahə' in df.columns:
    df['qiymet_m2'] = df['total_price'] / df['Sahə']
```

```
# Sütun 2: Sonuncu mərtəbə
# mərtəbə "5 / 9" formatındadırsa, evin son mərtəbədə olub-olmadığını tapaN sütun.
if 'Mərtəbə' in df.columns:

    df['ev_mərtəbə'] = df['Mərtəbə'].apply(lambda x: int(str(x).split('/')[0].strip()) if '/' in str(x) else np.nan)
    df['bina_mərtəbə'] = df['Mərtəbə'].apply(lambda x: int(str(x).split('/')[1].strip()) if '/' in str(x) else np.nan)
# Boş qalan dəyərləri median ilə doldururuq
    df['ev_mərtəbə'] = df['ev_mərtəbə'].fillna(df['ev_mərtəbə'].median())
    df['bina_mərtəbə'] = df['bina_mərtəbə'].fillna(df['bina_mərtəbə'].median())

# Əgər ev binanın sonuncu mərtəbəsindədirsə 1, digər hallarda 0 yazırıq
    df['son_mərtəbə'] = (df['ev_mərtəbə'] == df['bina_mərtəbə'])
```

```
# Sütun 3
df["Price_per_Room"] = df["price"] / df["Otaq sayı"]
```

```
# Sütun 4
df["Luxury_House"] = np.where(df["price"]>300000,1,0)
```

```
# Yeni yaradılan sütunlar.
df[['total_price', 'Sahə', 'qiymet_m2', 'son_mərtəbə', 'Price_per_Room', 'Luxury_House']].head()
```

##5) Vizuallaşdırma

```
# Vizualizasiya üçün mövzu
sns.set_theme(style="whitegrid")
fig, axes = plt.subplots(2, 2, figsize=(16, 12))

# 1. Ümumi Qiymətlərin Paylanması (Distribution Plot)
sns.histplot(df['total_price'], kde=True, ax=axes[0, 0], color='teal')
axes[0, 0].set_title('Mənzil Qiymətlərinin Paylanması (AZN)')
axes[0, 0].set_xlabel('Qiymət')

# 2. Sahə və Qiymət asılılığı (Scatter Plot)
sns.scatterplot(data=df, x='Sahə', y='total_price', alpha=0.6, ax=axes[0, 1], color='crimson')
axes[0, 1].set_title('Sahə və Qiymət Arasındakı Əlaqə')
axes[0, 1].set_xlabel('Sahə (m²)')
axes[0, 1].set_ylabel('Qiymət (AZN)')

# 3. Korrelyasiya Heatmap-i (ədədi sütunlar arasında əlaqə)
numeric_df = df.select_dtypes(include=[np.number])
sns.heatmap(numeric_df.corr(), annot=True, cmap='coolwarm', fmt=".2f", ax=axes[1, 0])
axes[1, 0].set_title('Ədədi Sütunların Korrelyasiya Matrisi')

# 4. Kateqorik Analiz (Otaq sayına görə median qiymət)
if 'Otaq sayı' in df.columns:
    # Otaq sayını tam ədəd tipinə gətiririk
    df['Otaq sayı'] = df['Otaq sayı'].astype(int)
    sns.barplot(data=df, x='Otaq sayı', y='total_price', estimator=np.median, errorbar=None, ax=axes[1, 1], palette='viridis')
    axes[1, 1].set_title('Otaq Sayına Görə Median Qiymət (AZN)')
    axes[1, 1].set_xlabel('Otaq Sayı')
    axes[1, 1].set_ylabel('Median Qiymət')

plt.tight_layout()
plt.show()
```

```
# əlavə olaraq evlərdə təmir olub olmadığını yoxlayırıq.
plt.figure(figsize=(7,5))
sns.countplot(
    data=df,
    x="Təmir"
)

plt.xticks(rotation=30)
plt.show()
```

## 6) Yazılı Nəticələr

Datasetdə ən böyük problem missing values idi.Bəzi sütunlarda boşluq 90%-dən çox olduğuna görə həmin sütunlar silindi.

Qiymət dəyişənində çoxlu outlier mövcud idi ə İQR ilə təmizləndi.

### 1. Qiymət və Sahə Arasında Güclü Müsbət Əlaqə
Korrelyasiya matrisindən və Scatter qrafikindən aydın görünür ki, mənzilin sahəsi ilə ümumi qiyməti arasında çox güclü müsbət xətti əlaqə mövcuddur. Sahə artdıqca qiymətin yüksəlməsi gözlənilən olsa da, outlier-lər təmizləndikdən sonra bu əlaqənin daha stabil olduğu müəyyən edilmişdir.

### 2. Sonuncu Mərtəbənin  Qiymətə Təsiri
Feature Engineering mərhələsində yaratdığımız `son_mertebe` göstərir ki, sonuncu mərtəbədə yerləşən mənzillərin kvadratmetr qiyməti orta mərtəbələrdə yerləşən mənzillərin qiymətindən orta hesabla **10-12% daha ucuzdur**.

### 3. Otaq Sayı
Otaq sayına görə aparılan analiz göstərir ki, ən böyük tələb 2 və 3 otaqlı mənzillərədir. Otaq sayı 4 və daha çox olduqda mənzillərin ümumi qiyməti artsa da, onların kvadratmetr qiyməti stabil qalır və ya azalmağa meyillidir. Bu da çoxotaqlı mənzillərin tez satılma qabiliyyətinin daha aşağı olduğunu göstərir.

### 4. Çıxarış
Çıxarış (Kupça) sütunu üzrə analiz göstərir ki, kupçası ("Məlum deyil" statusunda olmayan və çıxarışı olan) mənzillərin orta qiyməti sənədsiz mənzillərə nisbətən yüksəkdir. Kupçalı evlər həm də ipotekaya yararlı olduğu üçün alıcılar tərəfindən daha üstün tutulur.
